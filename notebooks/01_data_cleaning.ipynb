```python
import pandas as pd

csv_path = "/content/data/raw/TORONTO.AOC1.csv"

# Load with explicit options (good practice for big files)
df = pd.read_csv(
    csv_path,
    low_memory=False
)

df.shape, df.columns
```

```python
df.head()
```

```python
df.info()
```

```python
# Make column names lowercase with underscores
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

df.columns
```

```python
# Parse date
df["activitystartdate"] = pd.to_datetime(
    df["activitystartdate"],
    errors="coerce"
)

# Some times may be missing; fill with midnight
time_str = df["activitystarttime"].fillna("00:00:00")

# Convert "hh:mm:ss" to a timedelta
time_delta = pd.to_timedelta(time_str)

# Full datetime
df["activity_datetime"] = df["activitystartdate"] + time_delta

# Time-based features
df["year"] = df["activity_datetime"].dt.year
df["month"] = df["activity_datetime"].dt.month
df["dayofweek"] = df["activity_datetime"].dt.day_name()

df[["activitystartdate", "activitystarttime", "activity_datetime", "year", "month", "dayofweek"]].head()
```

```python
# Latitude/longitude already numeric, but enforce just in case
df["monitoringlocationlatitude"] = pd.to_numeric(
    df["monitoringlocationlatitude"], errors="coerce"
)
df["monitoringlocationlongitude"] = pd.to_numeric(
    df["monitoringlocationlongitude"], errors="coerce"
)

# Result value (measurement)
df["resultvalue"] = pd.to_numeric(df["resultvalue"], errors="coerce")

df[["resultvalue", "resultunit"]].describe(include="all")
```

```python
critical_cols = [
    "monitoringlocationname",
    "activity_datetime",
    "characteristicname",
    "resultvalue",
]

before = len(df)
df_clean = df.dropna(subset=critical_cols)
after = len(df_clean)

print(f"Rows before: {before:,}")
print(f"Rows after : {after:,}")
print(f"Removed    : {before - after:,}")
```

```python
# Example: focus on some common characteristics
focus_chars = [
    "Temperature, water",
    "Conductivity",
    "pH",
    "Escherichia coli",
]

df_clean = df_clean[df_clean["characteristicname"].isin(focus_chars)]

df_clean["characteristicname"].value_counts()
```

```python
import os

clean_dir = "/content/data/cleaned"
os.makedirs(clean_dir, exist_ok=True)

clean_path = os.path.join(clean_dir, "toronto_aoc1_clean.csv")
df_clean.to_csv(clean_path, index=False)

print("Saved cleaned file to:", clean_path)
!ls -lh /content/data/cleaned
```
